```python
import pandas as pd

df = pd.read_csv(r"C:\Users\OTABIL\Downloads\python\Kaggle Datasets\Mall_Customers.csv")

# Preview the data
print(df.head())

# Check shape
print(df.shape)

# Check missing values
print(df.isnull().sum())
```

```python
# Check  duplicates exists
print("Duplicates:", df.duplicated().sum())
```

```python
# Remove them
df = df.drop_duplicates()

# Confirm
print("Shape after removing duplicates:", df.shape)
```

```python
# Check unique values in Gender column
print(df["Gender"].unique())
# Standardize to lowercase
df["Gender"] = df["Gender"].str.lower()
```

```python
# Check current column names
print(df.columns)
```

```python

# Rename columns
df.columns = ["customer_id", "gender", "age", "annual_income", "spending_score"]

# Confirm
print(df.columns)
```

```python
# Check current data types
print(df.dtypes)
```

```python
df.to_csv("Mall_Customers_Cleaned.csv", index=False)
print("Cleaned file saved!")
```
